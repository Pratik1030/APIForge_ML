```python

from pathlib import Path
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# --------------------------------------------------
# 1. Paths and dataset
# --------------------------------------------------
BASE_DIR = Path.cwd()

# If the notebook is not in your ApiForge folder, uncomment:
# BASE_DIR = Path(r"C:\Users\Owner\OneDrive\Desktop\ApiForge")

DATA_FILE = BASE_DIR / "ml_data" / "synthetic_api_events.csv"
MODEL_DIR = BASE_DIR / "ml_data" / "model"
MODEL_PATH = MODEL_DIR / "usage_forecast_model.joblib"

if not DATA_FILE.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_FILE}")

MODEL_DIR.mkdir(parents=True, exist_ok=True)

# --------------------------------------------------
# 2. Load data and aggregate monthly requests
# --------------------------------------------------
df = pd.read_csv(DATA_FILE)

required = {"timestamp"}
if not required.issubset(df.columns):
    raise ValueError(f"Dataset must contain columns: {required}")

df["timestamp"] = pd.to_datetime(
    df["timestamp"], errors="coerce", utc=True
)
df = df.dropna(subset=["timestamp"])

# Count each event once if an event ID is available.
if "eventId" in df.columns:
    df = df.drop_duplicates(subset=["eventId"])

df["month"] = (
    df["timestamp"]
    .dt.tz_localize(None)
    .dt.to_period("M")
    .dt.to_timestamp()
)

monthly = (
    df.groupby("month")
      .size()
      .rename("requests")
      .sort_index()
      .asfreq("MS", fill_value=0)
)

print("Dataset:", DATA_FILE.name)
print("Total request events:", len(df))
print("Date range:", df["timestamp"].min(), "to", df["timestamp"].max())
print("Observed months:", (monthly > 0).sum())

if (monthly > 0).sum() < 3:
    raise ValueError(
        "Not enough historical months to train this seasonal model. "
        "At least 3 distinct months with requests are required. "
        "Use a dataset with a longer date range; do not treat zero-filled "
        "months as genuine training history."
    )

# --------------------------------------------------
# 3. Create trend + sine + cosine features
# --------------------------------------------------
history = monthly.rename_axis("ds").reset_index()
history.columns = ["ds", "y"]

history["t"] = np.arange(len(history))
history["month_num"] = history["ds"].dt.month

history["sin_season"] = np.sin(
    2 * np.pi * history["month_num"] / 12
)
history["cos_season"] = np.cos(
    2 * np.pi * history["month_num"] / 12
)

FEATURES = ["t", "sin_season", "cos_season"]
X = history[FEATURES]
y = history["y"].astype(float)

# --------------------------------------------------
# 4. Train model
# --------------------------------------------------
model = LinearRegression()
model.fit(X, y)

history["fitted"] = np.maximum(model.predict(X), 0)

residuals = y.to_numpy() - history["fitted"].to_numpy()
residual_std = float(np.std(residuals, ddof=1)) if len(y) > 1 else 0.0

if y.nunique() > 1:
    score = r2_score(y, history["fitted"])
else:
    score = None

mae = mean_absolute_error(y, history["fitted"])

# --------------------------------------------------
# 5. Forecast the next 6 months
# --------------------------------------------------
HORIZON = 6
last_t = int(history["t"].iloc[-1])
last_month = history["ds"].max()

future_dates = pd.date_range(
    start=last_month + pd.offsets.MonthBegin(1),
    periods=HORIZON,
    freq="MS"
)

future = pd.DataFrame({"ds": future_dates})
future["t"] = np.arange(last_t + 1, last_t + HORIZON + 1)
future["month_num"] = future["ds"].dt.month
future["sin_season"] = np.sin(2 * np.pi * future["month_num"] / 12)
future["cos_season"] = np.cos(2 * np.pi * future["month_num"] / 12)

future["predicted"] = np.maximum(
    model.predict(future[FEATURES]), 0
)

# Approximate residual-spread band, not a calibrated prediction interval.
future["lower"] = np.maximum(
    future["predicted"] - 1.96 * residual_std, 0
)
future["upper"] = np.maximum(
    future["predicted"] + 1.96 * residual_std, 0
)

# --------------------------------------------------
# 6. Save fitted model and forecasting metadata
# --------------------------------------------------
artifact = {
    "model": model,
    "features": FEATURES,
    "last_month": last_month,
    "last_t": last_t,
    "residual_std": residual_std,
    "horizon": HORIZON,
    "training_months": len(history),
    "target": "monthly_request_count",
    "seasonality_period_months": 12,
    "source_dataset": DATA_FILE.name,
}

joblib.dump(artifact, MODEL_PATH)

# --------------------------------------------------
# 7. Print results
# --------------------------------------------------
print("\nMODEL TRAINED AND SAVED")
print("Model path:", MODEL_PATH)
print("Training months:", len(history))
print("Training MAE:", round(mae, 2))
print("Training R²:", round(score, 4) if score is not None else "Undefined")
print("\nSix-month forecast:")
display(
    future[["ds", "predicted", "lower", "upper"]]
    .assign(ds=lambda x: x["ds"].dt.strftime("%Y-%m"))
    .round(2)
)

# --------------------------------------------------
# 8. Display prediction graph
# --------------------------------------------------
plt.figure(figsize=(13, 6))

plt.plot(
    history["ds"], history["y"],
    marker="o", label="Actual requests"
)
plt.plot(
    history["ds"], history["fitted"],
    linestyle="--", marker=".", label="Historical model fit"
)
plt.plot(
    future["ds"], future["predicted"],
    marker="o", linestyle="--", label="6-month forecast"
)
plt.fill_between(
    future["ds"],
    future["lower"],
    future["upper"],
    alpha=0.2,
    label="Approximate uncertainty band"
)

plt.axvline(
    history["ds"].max(),
    linestyle=":",
    label="Forecast starts"
)
plt.title("APIForge — Monthly API Usage Forecast")
plt.xlabel("Month")
plt.ylabel("Number of requests")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

from pathlib import Path
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    explained_variance_score
)

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================================================
# 1. CONFIGURATION
# ============================================================
BASE_DIR = Path.cwd()

# Uncomment if your notebook is not in the ApiForge root:
# BASE_DIR = Path(r"C:\Users\Owner\OneDrive\Desktop\ApiForge")

DATA_FILE = BASE_DIR / "ml_data" / "synthetic_api_events.csv"
MODEL_DIR = BASE_DIR / "ml_data" / "model"
MODEL_PATH = MODEL_DIR / "usage_forecast_model.joblib"

TEST_FRACTION = 0.20
FORECAST_DAYS = 180
FORGETTING_FACTOR = 0.995
PRIOR_STRENGTH = 10.0

MODEL_DIR.mkdir(parents=True, exist_ok=True)

if not DATA_FILE.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_FILE}")

# ============================================================
# 2. LOAD DATA AND CREATE DAILY REQUEST COUNTS
# ============================================================
raw = pd.read_csv(DATA_FILE, low_memory=False)

if "timestamp" not in raw.columns:
    raise ValueError("Dataset must contain a timestamp column.")

raw["timestamp"] = pd.to_datetime(
    raw["timestamp"], errors="coerce", utc=True
)
raw = raw.dropna(subset=["timestamp"]).copy()

if "eventId" in raw.columns:
    raw = raw.drop_duplicates(subset=["eventId"])

if raw.empty:
    raise ValueError("No valid timestamped events found.")

raw["day"] = (
    raw["timestamp"]
    .dt.tz_localize(None)
    .dt.normalize()
)

daily = (
    raw.groupby("day")
       .size()
       .rename("requests")
       .sort_index()
)

# Include zero-request days inside the observed range.
daily = daily.asfreq("D", fill_value=0)

print("Dataset:", DATA_FILE.name)
print("Valid unique events:", len(raw))
print("Observed date range:", daily.index.min().date(),
      "to", daily.index.max().date())
print("Daily observations:", len(daily))

# ============================================================
# 3. FEATURES: TREND + SINE + COSINE
# ============================================================
def make_features(dates, origin_date):
    dates = pd.DatetimeIndex(dates)
    elapsed_days = (dates - origin_date).days.astype(float)

    # Scale trend for numerical stability.
    t = elapsed_days / 30.0

    # Annual seasonality using day of year.
    day_of_year = dates.dayofyear.to_numpy(dtype=float)
    year_length = np.where(dates.is_leap_year, 366.0, 365.0)

    sin_season = np.sin(2 * np.pi * day_of_year / year_length)
    cos_season = np.cos(2 * np.pi * day_of_year / year_length)

    # Columns: intercept, trend, sine, cosine
    return np.column_stack([
        np.ones(len(dates)),
        t,
        sin_season,
        cos_season
    ])

# ============================================================
# 4. CHRONOLOGICAL TRAIN / TEST SPLIT
# ============================================================
n = len(daily)

if n < 5:
    raise ValueError(
        f"Only {n} daily observations. At least 5 are needed "
        "for a basic chronological train/test split."
    )

split_idx = int(n * (1 - TEST_FRACTION))
split_idx = min(max(split_idx, 3), n - 1)

train = daily.iloc[:split_idx]
test = daily.iloc[split_idx:]

origin_date = daily.index[0]

X_train = make_features(train.index, origin_date)
X_test = make_features(test.index, origin_date)

y_train = train.to_numpy(dtype=float)
y_test = test.to_numpy(dtype=float)

print("\nTraining days:", len(train))
print("Testing days:", len(test))
print("Train period:", train.index.min().date(), "to",
      train.index.max().date())
print("Test period:", test.index.min().date(), "to",
      test.index.max().date())

# ============================================================
# 5. COLD-START RECURSIVE LEAST SQUARES
# ============================================================
# Prior:
# - baseline starts at the training average
# - trend starts at zero
# - seasonal effects start at zero
#
# Observed training data then update the weights sequentially.
# PRIOR_STRENGTH controls how strongly the initial prior is retained.

feature_count = X_train.shape[1]

theta = np.array([
    max(float(np.mean(y_train)), 0.0),
    0.0,
    0.0,
    0.0
])

P = np.eye(feature_count) * PRIOR_STRENGTH
P[0, 0] = max(float(np.var(y_train)), 1.0) * PRIOR_STRENGTH

def rls_update(theta, P, x, y, forgetting_factor=0.995):
    x = np.asarray(x, dtype=float).reshape(-1, 1)

    denominator = forgetting_factor + float(x.T @ P @ x)
    gain = (P @ x) / denominator

    prediction = float((x.T @ theta.reshape(-1, 1))[0, 0])
    error = float(y - prediction)

    theta_new = theta + (gain[:, 0] * error)
    P_new = (P - gain @ x.T @ P) / forgetting_factor

    # Numerical stabilization
    P_new = (P_new + P_new.T) / 2

    return theta_new, P_new

# Fit only on the training period.
for x, y in zip(X_train, y_train):
    theta, P = rls_update(
        theta, P, x, y,
        forgetting_factor=FORGETTING_FACTOR
    )

# Freeze the model for an honest holdout evaluation.
test_predictions_raw = X_test @ theta
test_predictions = np.maximum(test_predictions_raw, 0)

# Historical training predictions from the final trained weights.
train_predictions = np.maximum(X_train @ theta, 0)

# ============================================================
# 6. EVALUATION METRICS
# ============================================================
def calculate_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(actual, predicted)
    rmse = float(np.sqrt(mean_squared_error(actual, predicted)))

    nonzero = actual != 0
    mape = (
        float(np.mean(
            np.abs((actual[nonzero] - predicted[nonzero])
                   / actual[nonzero])
        ) * 100)
        if nonzero.any() else None
    )

    denominator = np.sum(np.abs(actual))
    wape = (
        float(np.sum(np.abs(actual - predicted)) / denominator * 100)
        if denominator > 0 else None
    )

    smape_denominator = np.abs(actual) + np.abs(predicted)
    valid = smape_denominator != 0
    smape = (
        float(np.mean(
            2 * np.abs(actual[valid] - predicted[valid])
            / smape_denominator[valid]
        ) * 100)
        if valid.any() else None
    )

    return {
        "MAE": float(mae),
        "RMSE": rmse,
        "R2": (
            float(r2_score(actual, predicted))
            if len(actual) >= 2 and np.var(actual) > 0
            else None
        ),
        "Explained variance": (
            float(explained_variance_score(actual, predicted))
            if len(actual) >= 2 else None
        ),
        "MAPE (%)": mape,
        "WAPE (%)": wape,
        "sMAPE (%)": smape,
        "Mean error (bias)": float(np.mean(predicted - actual))
    }

train_metrics = calculate_metrics(y_train, train_predictions)
test_metrics = calculate_metrics(y_test, test_predictions)

metrics_df = pd.DataFrame({
    "Training": train_metrics,
    "Testing": test_metrics
})

print("\nEVALUATION METRICS")
display(metrics_df.round(4))

# ============================================================
# 7. FORECAST NEXT 180 DAYS
# ============================================================
last_day = daily.index.max()
future_dates = pd.date_range(
    start=last_day + pd.Timedelta(days=1),
    periods=FORECAST_DAYS,
    freq="D"
)

X_future = make_features(future_dates, origin_date)
future_daily_predictions = np.maximum(X_future @ theta, 0)

forecast_daily = pd.DataFrame({
    "date": future_dates,
    "predicted_requests": future_daily_predictions
})

forecast_daily["month"] = (
    forecast_daily["date"].dt.to_period("M").astype(str)
)

forecast_monthly = (
    forecast_daily.groupby("month", as_index=False)
    ["predicted_requests"].sum()
)

print("\nNEXT SIX MONTHS — FORECAST REQUEST COUNTS")
display(forecast_monthly.round(0))

# ============================================================
# 8. SAVE MODEL AND METADATA
# ============================================================
artifact = {
    "model_type": "Recursive Least Squares with cold-start prior",
    "theta": theta,
    "covariance": P,
    "features": ["intercept", "trend_months", "sin_annual", "cos_annual"],
    "origin_date": origin_date,
    "last_observed_day": last_day,
    "forgetting_factor": FORGETTING_FACTOR,
    "prior_strength": PRIOR_STRENGTH,
    "training_days": len(train),
    "testing_days": len(test),
    "training_metrics": train_metrics,
    "testing_metrics": test_metrics,
    "forecast_horizon_days": FORECAST_DAYS,
    "source_dataset": DATA_FILE.name,
    "note": (
        "Seasonal weights are weakly identified from January-only data. "
        "Future real observations should update this artifact."
    )
}

joblib.dump(artifact, MODEL_PATH)

print("\nSaved model:", MODEL_PATH)

# Save evaluation results and forecasts for later use.
metrics_df.to_csv(MODEL_DIR / "usage_model_metrics.csv")
forecast_daily.to_csv(
    MODEL_DIR / "usage_daily_forecast.csv", index=False
)
forecast_monthly.to_csv(
    MODEL_DIR / "usage_monthly_forecast.csv", index=False
)

# ============================================================
# 9. GRAPH: TRAIN, TEST, AND FUTURE FORECAST
# ============================================================
plt.figure(figsize=(14, 6))

plt.plot(
    train.index, y_train,
    label="Training actuals", marker=".", linewidth=1
)
plt.plot(
    test.index, y_test,
    label="Test actuals", marker="o", linewidth=1
)
plt.plot(
    test.index, test_predictions,
    label="Test predictions", marker="o", linestyle="--"
)
plt.plot(
    future_dates, future_daily_predictions,
    label="Future forecast", linestyle="--"
)

plt.axvline(
    test.index[0],
    linestyle=":",
    label="Test starts"
)
plt.axvline(
    future_dates[0],
    linestyle=":",
    label="Forecast starts"
)

plt.title("APIForge Usage Prediction — Cold-Start RLS")
plt.xlabel("Date")
plt.ylabel("Requests per day")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Separate monthly forecast chart
plt.figure(figsize=(11, 5))
plt.bar(
    forecast_monthly["month"],
    forecast_monthly["predicted_requests"]
)
plt.title("APIForge — Forecast Monthly Request Volume")
plt.xlabel("Month")
plt.ylabel("Predicted requests")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
